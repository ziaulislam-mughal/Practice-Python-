<<<_OPERATORS IN PYTHON_>>>>
   Operators in Python are symbols or keywords used to perform operations on variables and values like calculations, comparisons, or logical decisions.

TYPES OF OPERATORS

1. Arithmetic Operators:  
Used for mathematical calculations.
(+ - * / % ** //)

```python
a = 9
b = 2

print(a + b)  # Output: 11
print(a - b)  # Output: 7
print(a * b)  # Output: 18
print(a / b)  # Output: 4.5
print(a % b)  # Output: 0 reminder 
#power 
print(a ** b)
#floor division
print(a // b)  # Output: 4
```

2. Assingment operators:

Used to assign or update values.
(= += -= *= /=)


```python
x = 5
y = 10

y += 5 # y = y + 5 
print(y)  # Output: 15

x *= 2 # x = x * 2
print(x)
```

3. Comparison (Relational Operators ):

Compare two values and return True or False.
(== != > < >= <=)



```python
a = 11
b = 3 

print( a == b )
print( a != b )
print( a > b )
print(a < b)
print( a >= b )
print(a <= b)
```

4.Logical Operators:

Used to combine conditions.
(and or not)


```python
4. # Logical Operators and, or, not
# and -> multiplication -> 0 - > false ,1 -> true 
# or -> addition 
# not -> negation

p = 10
q = 4 

print ( (p > 5) and (q < 3) )  # True and false  -> false
print ( (p > 5) or (q < 3) )   # True or false  -> true
print ( not(p > 5) )            # not true -> false
```

5. Membership operatos:

Checks if a value exists in a sequence i.e; list string, etc.
(in not in)


```python
# Membership Operators

fruits = [ "banana", "cherry"]
```

```python
print("apple"  in fruits)  
```
