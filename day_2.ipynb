Operators in python 

1. Arithmetic Operators 

```python
a = 9
b = 2

print(a + b)  # Output: 11
print(a - b)  # Output: 7
print(a * b)  # Output: 18
print(a / b)  # Output: 4.5
print(a % b)  # Output: 0 reminder 
#power 
print(a ** b)
#floor division
print(a // b)  # Output: 4
```

```python
# Example: PEMDAS Rule in Python


# PEMDAS stands for:
# Parentheses
# Exponents
# Multiplication and Division
# Addition and Subtraction

result = 5 + 3 * 2 ** 2 - (4 / 2)

print("Expression: 5 + 3 * 2 ** 2 - (4 / 2)")
print("Result:", result)

# Step-by-step explanation:
# 2 ** 2 = 4
# 3 * 4 = 12
# (4 / 2) = 2.0
# 5 + 12 - 2.0 = 15.0
```

2. Assingment operators

```python
x = 5
y = 10

y += 5 # y = y + 5 
print(y)  # Output: 15

x *= 2 # x = x * 2
print(x)
```

```python
# Advanced Assignment Operator Examples


x = 10
y = 4

print("Initial Values:")
print("x =", x)
print("y =", y)

# Adding y + 4 to x
x += y + 4
print("\nAfter x += y + 4")
print("x =", x)

# Multiplying x by (y - 2)
x *= y - 2
print("\nAfter x *= y - 2")
print("x =", x)

# Dividing x by (y + 1)
x /= y + 1
print("\nAfter x /= y + 1")
print("x =", x)

print("\nFinal Data Type of x:", type(x))
```

3. Comparison (Relational Operators )

```python
a = 11
b = 3 

print( a == b )
print( a != b )
print( a > b )
print(a < b)
print( a >= b )
print(a <= b)
```

```python
4. # Logical Operators and, or, not
# and -> multiplication -> 0 - > false ,1 -> true 
# or -> addition 
# not -> negation

p = 10
q = 4 

print ( (p > 5) and (q < 3) )  # True and false  -> false
print ( (p > 5) or (q < 3) )   # True or false  -> true
print ( not(p > 5) )            # not true -> false
```

5. Membership operatos in, not in

```python
# Membership Operators

fruits = [ "banana", "cherry"]
```

```python
print("apple"  in fruits)  
```

```python
# Example: Membership Operators (in / not in)


# Example with string
name = "Abdul Rafhy"

print("Is 'Rafhy' in name?", "Rafhy" in name)
print("Is 'Ali' not in name?", "Ali" not in name)

# Example with list
numbers = [10, 20, 30, 40]

print("Is 20 in numbers?", 20 in numbers)
print("Is 50 not in numbers?", 50 not in numbers)

# Example with tuple
vegetables= ("carrot", "spanish", "cucumber")

print("Is 'cucumber' in vegetables?", "cucumber" in vegetables)
```

# Mini Practice Program

```python


# Program 2: Simple Discount Calculator


# Taking user input
price = float(input("Enter product price: "))
discount_percent = float(input("Enter discount percentage: "))

# Arithmetic operator
discount_amount = (price * discount_percent) / 100

# Assignment operator with expression
price -= discount_amount + 2  # extra 2 currency service fee adjustment

print("\nPrice after discount and adjustment:", price)

# Comparison operator
is_expensive = price > 100

# Logical operator
special_offer = price > 50 and price < 150

# Membership operator
available_categories = ["electronics", "clothing", "books"]
category = input("Enter product category: ")

is_valid_category = category in available_categories

print("\n------ Summary ------")
print("Is price greater than 100?", is_expensive)
print("Is price between 50 and 150?", special_offer)
print("Is category valid?", is_valid_category)
```
