```python
string = "aabbcc"
count = 3 
while True :
    if string[0] == 'a':
        print("if statment")
        string = string[2:]
    elif string[-1] == 'c':
        print("elif statment")
        string = string[:2]
    else :
        print("else statment")
        count +=1 
        break

print(count)
print(string)
```
