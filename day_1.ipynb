## ******************* Variable ********************

### Variable:
   -> A variable in Python is a named storage location that holds a value, which can be changed during program execution.It acts as a container for data.  

   -> Syntax to create a variable: variable_name = value

```python
# Example:

# Note:Choose meaningful English names that clearly describe the variable’s purpose.

student_age = 25 
print(f"Age of Student is: {student_age}")
```

********************************************************************************************

### Rules of Variable Naming:

-> Three rules for naming variables in Python:  
      1.Variable names must start with a letter, or an underscore(_), not a number.  
      2.Variable name can contain letters, numbers, and underscore, but no special characters like @, #, or spaces.  
      3.Variable names are case-sensitive (e.g., "Age" and "age" are different).

## ******************* Datatypes ********************



-> In Python, data types define the kind of value a variable can hold.   
* Main types:
   - integer
   - float
   - string
   - boolean


     

*****************************************************************************************************

### Integer Data Type:  
 -> An integer data type (int) represents whole numbers without a decimal point, such as positive, negative, or zero values.


```python
#Example:
student_marks = 90
print(type(student_marks))
print(f"Student Marks = {student_marks}")
```

**********************************************************************************

### Float Data Type:
-> A float data type represents numbers with a decimal point or fractional part, used for real numbers.

```python
#Example:
person_height = 5.7
print(type(person_height))
print(f"Height of a person is: {person_height}")
```

************************************************************************************

### String Data Type:
-> A string data type (str) represents a sequence of characters, such as text,enclosed in quotes.It can include letters, numbers, symbols, or spaces.

* Two ways to create a string:  
   -  Using double quotes.  
   -  Using multi-line string (triple quotes).

```python
#Example:

#************* Using Duoble Quotes ***************

student_name = "Sophia"
print(type(student_name))
print(f"Student name: {student_name}")
print("------------------------")

#************* Using Triple Quotes ***************

student_info = """Name: Sophia 
Class: 10th
Section: A"""
print(type(student_info))
print("Student Information:")
print(student_info)
```

*******************************************************************************************

### Boolean Data Type:
-> A boolean data type (bool) represents truth values, used for logical operations and conditions.
* Possible values: 
   - True or False. 

```python
#Example:
is_student = True
print(type(is_student))
print(is_student)
```

## *********** Python Built-in Functions  ************


-> Python Built-in Functions are pre-defined functions in Python that help perform common tasks without importing anything.
* Main types:
  - type()
  - input
  - output

*******************************************************************************************

### type() Function:
-> The type() function in Python returns the data type of a variable or value, helping to identify what kind of data it holds.

```python
#Example:
person_age = 20
print(type(person_age))
```

*****************************************************************************************

### input() Function:
-> The input() function is used in Python to take user input from the keyboard during program execution. 

-> The default data type of the input is string(str).

```python
#Example 1:
user_name = input("Enter your name:")
print(type(user_name))
print(user_name)
```

```python
#Example 1:

# input() returns values as STRINGS by default

a = input("First number: ")
b = input("Second number: ")
print("Without int:", a + b)

# 23 appears because "2" and "3" are strings
# Python joins the text instead of adding numbers
# "2" + "3" = "23"  (this is called concatenation)

print("-----------------------------")

# int() converts input to integers for real addition

a = int(input("First number: "))
b = int(input("Second number: "))
print("With int:", a + b)

# Now values are integers, so 2 + 3 = 5
```

*************************************************************************************************

### print() function:
-> The print() function is a built-in Python function used to display output on the screen.
It shows text, numbers, or variable values to the user.

```python
#Example:

name = "Sophia"
print(name)
```

********************************************************************************************

### Type Casting:
-> Type casting is the process of converting a value from one data type to another, such as changing a string to an integer or float.

```python
#Example:

#**************** String age ********************

age_str = "25"
print(type(age_str))
print(f"String age: {age_str}")

#**************** Integer age ********************

print("--------------------")
age_int = int(age_str)
print(type(age_int))
print(f"Integer age: {age_int}")
```

***************************************************************************************************

## ************** formatted string ***************


-> A formatted string allows you to insert variable values inside a string easily using {} brackets and an f before the quotes.

```python
name = "Sophia"
age = 20

# using f-string

print(f"My name is {name} and I am {age} years old.")
```

********************************************************************************************

### ********** Multiple Variable Assignment in Python **********

-> In Python, you can assign values to multiple variables at once using comma-separated variables and values.

-> This method makes the code shorter and cleaner instead of assigning values in separate lines.

```python
# multiple assignment

x, y, z = 1, 2, 3

print(x)  # output: 1
print(y)  # output: 2
print(z)  # output: 3
```
