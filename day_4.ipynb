```python
#range(start,end,step)
for i in range(1,10,2):
    print(i)
```

```python
num = int(input("Enter a number : "))
for i in range(1,11):
    print(num , "x" , i , "=" , num*i)
    # 2 x 1 = 2
    # 2 x 2 = 4
```

```python
name = "Uajk king abdullah campus university"
for i in range(5+1) :
    print(i) 
    print(name)
```

```python
# while loop 
num = 1 # 2  
while num <= 5 :
    print(num)
    num += 1 # num = num + 1 -> 1 = 1 + 1 = 2 -> num = 2 
```

```python
while True :
    print("-"*45)
    print("1. Say Hellow ")
    print("2. Say Goodbye ")
    print("3. Exit ")

    choice = int(input("Enter your choice : "))

    if choice == 1 :
        print("-"*45)
        print("Hellow") 
        print("-"*45)
    elif choice == 2 :
        print("-"*45)
        print("Goodbye")
        print("-"*45)
    elif choice == 3 :
        print("-"*45)
        print("Exiting...")
        print("-"*45)
        break
    else: 
        print("Invalid choice. Please try again.")
```
