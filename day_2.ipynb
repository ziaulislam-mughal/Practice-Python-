## 1. Arithmetic Operators:
-> Arithmetic operators in Python are symbols used to perform basic mathematical operations like addition, subtraction, multiplication, and division on numbers.

| Operator | Meaning             | Example  | Output |
| -------- | ------------------- | -------- | ------ |
| `+`      | Addition            | `5 + 3`  | `8`    |
| `-`      | Subtraction         | `5 - 3`  | `2`    |
| `*`      | Multiplication      | `5 * 3`  | `15`   |
| `/`      | Division            | `5 / 2`  | `2.5`  |
| `//`     | Floor Division      | `5 // 2` | `2`    |
| `%`      | Modulus (remainder) | `5 % 2`  | `1`    |
| `**`     | Exponent            | `5 **2`  | `25`   |


```python
# Example:

a = 10
b = 3

# Addition: adds a and b
print("Addition: ",a + b)

# Subtraction: subtracts b from a
print("Subtraction: ",a - b)

# Multiplication: multiplies a and b
print("Multiplication:",a * b)

# Division: divides a by b (gives decimal result)
print("Division:",a / b)

# Floor Division: divides a by b and gives only whole number part
print("Floor Division:",a // b)

# Modulus: gives the remainder of a divided by b
print("Modulus:",a % b)

# Exponent: raises a to the power of b
print("Exponent:",a ** b)
```

## 2. Assingment operators:
-> Assignment operators in Python are used to store values in variables and update them efficiently.

| Operator | Meaning                 | Example   | Explanation          |
| -------- | ----------------------- | --------- | -------------------- |
| `=`      | Simple assignment       | `x = 5`   | Assigns 5 to x       |
| `+=`     | Add and assign          | `x += 3`  | Same as `x = x + 3`  |
| `-=`     | Subtract and assign     | `x -= 2`  | Same as `x = x - 2`  |
| `*=`     | Multiply and assign     | `x *= 4`  | Same as `x = x * 4`  |
| `/=`     | Divide and assign       | `x /= 2`  | Same as `x = x / 2`  |
| `//=`    | Floor divide and assign | `x //= 3` | Same as `x = x // 3` |
| `%=`     | Modulus and assign      | `x %= 2`  | Same as `x = x % 2`  |
| `**=`    | Exponent and assign     | `x **= 3` | Same as `x = x ** 3` |


```python
# Example:

# Starting value
x = 5
print("Initial value:", x)  # Output: 5

# Add and assign
x += 3
print("After x += 3:", x)   # Output: 8

# Subtract and assign
x -= 2
print("After x -= 2:", x)   # Output: 6

# Multiply and assign
x *= 4
print("After x *= 4:", x)   # Output: 24

# Divide and assign
x /= 2
print("After x /= 2:", x)   # Output: 12.0

# Floor divide and assign
x //= 5
print("After x //= 5:", x)  # Output: 2.0

# Modulus and assign
x %= 2
print("After x %= 2:", x)   # Output: 0.0

# Exponent and assign
x **= 3
print("After x **= 3:", x)  # Output: 0.0
```

## 3. Comparison (Relational Operators):
-> Comparison operators in Python are used to compare two values and return a Boolean result (True or False).

| Operator | Meaning               | Example  | Result |
| -------- | --------------------- | -------- | ------ |
| `==`     | Equal to              | `5 == 5` | True   |
| `!=`     | Not equal to          | `5 != 3` | True   |
| `>`      | Greater than          | `5 > 3`  | True   |
| `<`      | Less than             | `5 < 3`  | False  |
| `>=`     | Greater than or equal | `5 >= 5` | True   |
| `<=`     | Less than or equal    | `5 <= 3` | False  |


```python
#Example:

a = 5
b = 3

print(a == b)   # Equal to? False
print(a != b)   # Not equal to? True
print(a > b)    # Greater than? True
print(a < b)    # Less than? False
print(a >= b)   # Greater than or equal? True
print(a <= b)   # Less than or equal? False
```

## Logical Operator:
-> Logical operators in Python are used to combine multiple conditions and return a Boolean result (True or False).

| Operator | Meaning                                | Example                  |
| -------- | -------------------------------------- | ------------------------ |
| `and`    | True if both conditions are True       | `True and False` → False |
| `or`     | True if at least one condition is True | `True or False` → True   |
| `not`    | Reverses the Boolean value             | `not True` → False       |


```python
#Example:

x = 5
y = 10

# and: True only if both conditions are True
print(x > 2 and y > 5)   # True, because both conditions are True

# or: True if at least one condition is True
print(x > 2 or y < 5)    # True, because first condition is True

# not: reverses the Boolean value
print(not(x > 2))        # False, because x > 2 is True, not True = False
```

## 5. Membership Operatos:
->Membership operators in Python are used to check whether a value exists (or does not exist) in a sequence, returning True or False.

| Operator | Meaning                                  | Example                     |
| -------- | ---------------------------------------- | --------------------------- |
| `in`     | True if value is present in sequence     | `"a" in "apple"` → True     |
| `not in` | True if value is NOT present in sequence | `"b" not in "apple"` → True |


```python
# Example:

fruits = ["apple", "banana", "mango"]

# in: check if value exists in sequence
print("apple" in fruits)      # True, because "apple" is in the list

# not in: check if value does not exist in sequence
print("grape" not in fruits)  # True, because "grape" is not in the list
```
