```python
rows = 5
k = 2 * rows - 2
for i in range(rows, -1, -1):
    for j in range(k, 0, -1):
        print(end=" ")
    k = k + 1
    for j in range(0, i + 1):
        print("*", end=" ")
    print(" ")
```

```python
while True:
    a = float(input("ENter number "))
    b = float(input("ENter number "))
    operation = input("Enter operation  , or 'exit' to quit: ")

    if operation.lower() == "exit" or operation.lower() == "e":
        print("Good Bye")
        break

    if operation == "+":
        print(a + b)
    elif operation == "-":
        print(a - b)
    elif operation == "*":
        print(a * b)
    elif operation == "/":
        if b != 0:
            print(a / b)
        else:
            print("Error: Division by zero")
    else:
        print("Invalid operation")
        break
```

```python
def video_transcript(video_id):
    from youtube_api import youtubeapi


    yt = youtubeapi()
    transcript_dict = yt.fetch(video_id, languages=["eng"])
    transcript = " ".join(chunk.text for chunk in transcript_dict)
    print(transcript)
```
